## A. O R como calculadora
O R faz contas como uma calculadora. As operações básicas são:

| Operação | Símbolo | Exemplo | Resultado |
|---|---|---|---|
| Soma | `+` | `5 + 2` | 7 |
| Subtração | `-` | `5 - 2` | 3 |
| Multiplicação | `*` | `5 * 2` | 10 |
| Divisão | `/` | `5 / 2` | 2.5 |
| Potência | `^` | `5 ^ 2` | 25 |
| Divisão inteira | `%/%` | `5 %/% 2` | 2 |
| Resto da divisão | `%%` | `5 %% 2` | 1 |

Atenção: o R usa **ponto** como separador decimal (`2.5`), e não vírgula. A vírgula separa os argumentos das funções, como veremos adiante.

**Comentários.** Nas células de código, tudo o que vem depois de `#` numa linha é um **comentário**: o R ignora esse texto, que serve para explicar o código a quem lê, inclusive a você mesmo daqui a alguns meses. Comentar o código é parte da documentação da análise.

````r
# Cada linha é executada em ordem, e o resultado de cada uma aparece abaixo da célula
5 + 2
5 - 2
5 * 2
5 / 2
5 ^ 2
5 %/% 2   # quantas vezes o 2 cabe inteiro no 5
5 %% 2    # o que sobra dessa divisão
````

A **ordem das operações** é a mesma da matemática: primeiro potências, depois multiplicações e divisões, por último somas e subtrações. Na dúvida, use parênteses: eles deixam a intenção clara para quem lê.

````r
2 + 3 * 4     # a multiplicação vem primeiro: 2 + 12
(2 + 3) * 4   # os parênteses mudam a ordem: 5 * 4
````

Operações mais elaboradas são feitas com **funções**: o nome da função seguido de parênteses, com os valores dentro. Algumas funções matemáticas comuns:

````r
sqrt(16)       # raiz quadrada
abs(-3)        # valor absoluto (módulo)
exp(1)         # número de Euler elevado a 1
log(100)       # ATENÇÃO: log() é o logaritmo NATURAL (base e), não o de base 10
log10(100)     # logaritmo de base 10
log2(8)        # logaritmo de base 2
round(3.14159, digits = 2)   # arredonda para 2 casas decimais
````

> **Cuidado com o `log()`.** Em muitas calculadoras e no Excel, "log" é o logaritmo de base 10. No R, `log()` é o logaritmo natural. Esse é um erro comum ao refazer no R uma análise feita em outro programa.

Até aqui, cada resultado apareceu na tela e se perdeu. Para guardar um resultado e usá-lo depois, damos um **nome** a ele com a seta `<-`. Isso se chama **atribuição**, e o valor guardado com um nome é um **objeto**.

Leia `peso <- 250` como "peso recebe 250".

````r
peso <- 250   # guarda o valor 250 no objeto "peso" (nada aparece na tela)
peso          # para ver o conteúdo, basta rodar o nome do objeto
````

Algumas regras e boas práticas para os nomes:

- Prefira nomes que digam o que o objeto guarda: `dose_mg` é melhor que `x`.
- `=` também faz atribuição (`peso = 250`), mas a seta `<-` é a convenção do R e é a forma usada neste notebook.

## D. Tipos de dados
Cada valor no R tem um **tipo**, que define o que se pode fazer com ele. Os mais comuns são:

| Tipo | O que guarda | Exemplos |
|---|---|---|
| `numeric` | números, inteiros ou decimais | `250`, `3.5`, `-1` |
| `character` | texto, sempre entre aspas | `"A"`, `"controle"`, `"10 mg/kg"` |
| `logical` | verdadeiro ou falso | `TRUE`, `FALSE` |

Há também o tipo `integer` (números inteiros, escritos com `L`, como `5L`), que aparece com menos frequência no dia a dia. A função `class()` mostra o tipo de um valor.

````r
class(250)
class(3.5)
class("controle")
class(TRUE)
````

O tipo importa porque **as operações dependem dele**. Um número digitado entre aspas é texto, e texto não entra em contas:

````r
"10"        # entre aspas: é um texto, não um número
class("10")

# A linha abaixo daria erro ("non-numeric argument to binary operator"),
# porque não é possível somar texto e número. Tire o # do início para ver o erro:
# "10" + 1

as.numeric("10") + 1   # convertendo o texto em número, a conta funciona
````

Isso é comum ao importar planilhas: uma única célula com texto, como "8,5" (com vírgula) ou "não medido", faz a coluna inteira ser lida como texto. Quando a conversão não é possível, o R não inventa um valor: devolve `NA` e emite um aviso (*warning*).

````r
as.numeric("8.5")         # funciona: 8.5
as.numeric("8,5")         # não funciona: vírgula não é separador decimal no R → NA, com aviso
as.numeric("não medido")  # não funciona → NA, com aviso
````

### Dados ausentes: `NA`
`NA` (*not available*) representa um **valor ausente**: um animal que morreu antes da medida, uma amostra perdida, uma célula em branco. O R não ignora o `NA` por conta própria, e qualquer conta que envolva um valor desconhecido dá um resultado desconhecido:

````r
glicemia <- c(98, 105, NA, 110)   # c() junta valores; veremos isso na seção F

mean(glicemia)                 # a média de valores com um desconhecido é desconhecida: NA
mean(glicemia, na.rm = TRUE)   # na.rm = TRUE: remove os NA antes de calcular
is.na(glicemia)                # quais valores estão ausentes?
sum(is.na(glicemia))           # quantos valores estão ausentes?
````

> **Por que o R não remove os `NA` sozinho?** Porque essa é uma decisão de análise. Com `na.rm = TRUE`, a exclusão fica escrita no código, e qualquer pessoa que leia a análise sabe que ela foi feita. Na planilha, uma célula vazia pode ser ignorada sem que ninguém perceba.

## E. Comparações e operadores lógicos
Comparações respondem com `TRUE` (verdadeiro) ou `FALSE` (falso). Elas são a base para **filtrar dados**, como faremos na Parte 2 com `filter(dose == 1)`.

| Operador | Significado |
|---|---|
| `==` | igual a (dois sinais de igual: um só faz atribuição) |
| `!=` | diferente de |
| `<`, `>` | menor que, maior que |
| `<=`, `>=` | menor ou igual, maior ou igual |
| `&` | E: as duas condições são verdadeiras |
| <code>&#124;</code> | OU: pelo menos uma condição é verdadeira |
| `!` | NÃO: inverte o resultado |
| `%in%` | está contido em um conjunto de valores |

````r
pressao_animal <- 172

pressao_animal > 140                              # é hipertenso? (limiar ilustrativo)
pressao_animal == 172                             # é igual a 172?
pressao_animal != 172                             # é diferente de 172?
pressao_animal > 140 & pressao_animal < 180       # está entre 140 e 180? (E)
pressao_animal < 100 | pressao_animal > 160       # está fora da faixa 100–160? (OU)
!(pressao_animal > 140)                           # NÃO é maior que 140?
"A" %in% c("A", "B")                              # "A" está entre os fármacos estudados?
````

> **Uma armadilha com números decimais.** O computador guarda números decimais com uma pequena imprecisão. Por isso, comparar decimais com `==` pode dar um resultado inesperado. Para comparar decimais, use `all.equal()`, que tolera essas pequenas diferenças.

````r
0.1 + 0.2 == 0.3            # FALSE! 0.1 + 0.2 vale 0.30000000000000004 no computador
isTRUE(all.equal(0.1 + 0.2, 0.3))   # TRUE: compara com uma tolerância
````

## F. Vetores
Até aqui, cada objeto guardou um único valor. Na análise de dados, porém, trabalhamos com **conjuntos de valores**: as pressões de 10 animais, as doses de um experimento. A estrutura básica para isso é o **vetor**, uma sequência de valores **do mesmo tipo**, criada com a função `c()` (de *combinar*).

````r
pas <- c(172, 181, 165, 178, 169)   # pressão arterial sistólica (mmHg) de 5 animais
pas

length(pas)   # quantos valores o vetor tem
````

````r
# Outras formas de criar vetores
1:5                             # sequência de 1 a 5
seq(0, 120, by = 30)            # de 0 a 120, de 30 em 30 (ex.: tempos de coleta, em minutos)
rep("A", times = 3)             # repete um valor
c("controle", "tratado")        # vetor de texto
````

### Acessar elementos: colchetes `[ ]`
Cada elemento tem uma posição, e **a contagem começa em 1** (em algumas linguagens, como o Python, começa em 0).

````r
pas[1]          # o primeiro elemento
pas[c(1, 3)]    # o primeiro e o terceiro
pas[2:4]        # do segundo ao quarto
pas[-1]         # todos, menos o primeiro
````

Também é possível selecionar elementos com uma **condição**: dentro dos colchetes, um vetor de `TRUE`/`FALSE` mantém só as posições com `TRUE`. É exatamente o que a função `filter()` faz com as linhas de uma tabela, na Parte 2.

````r
pas > 170         # quais animais têm pressão acima de 170? (um TRUE/FALSE para cada animal)
pas[pas > 170]    # os valores que atendem à condição
````

## G. Vetorização: uma operação, todos os elementos
No R, a maioria das operações é **vetorizada**: aplicada a um vetor, ela age sobre **cada elemento**, sem que seja preciso repetir a conta valor por valor. É o equivalente a escrever uma fórmula numa célula da planilha e arrastá-la para a coluna inteira, só que sem arrastar e sem o risco de a fórmula parar uma linha antes do fim.

Um exemplo: converter a glicemia de mg/dL para mmol/L (divide-se por cerca de 18).

````r
glicemia_mg_dl <- c(98, 105, 142, 110, 187)

glicemia_mmol_l <- glicemia_mg_dl / 18   # a divisão é feita para cada um dos 5 valores
round(glicemia_mmol_l, 1)                # round() também é vetorizada
````

Operações entre **dois vetores do mesmo tamanho** são feitas **elemento a elemento**: o primeiro com o primeiro, o segundo com o segundo, e assim por diante.

````r
# Pressão de 4 animais antes e depois do tratamento
pas_antes  <- c(180, 175, 190, 168)
pas_depois <- c(162, 170, 171, 165)

reducao <- pas_antes - pas_depois   # a redução de cada animal
reducao
mean(reducao)                       # a redução média
````

Comparações também são vetorizadas, e isso permite contar e calcular proporções com uma linha. Para o R, `TRUE` vale 1 e `FALSE` vale 0:

````r
reducao > 10          # cada animal teve redução maior que 10 mmHg?
sum(reducao > 10)     # quantos animais tiveram (soma dos TRUE)
mean(reducao > 10)    # que proporção dos animais teve
````

> **Cuidado: vetores de tamanhos diferentes.** Se os vetores não têm o mesmo tamanho, o R **recicla** o mais curto: repete seus valores até completar o mais longo. Quando um tamanho não é múltiplo do outro, o R avisa, mas, quando é múltiplo, a reciclagem acontece em silêncio, o que pode esconder um erro (por exemplo, um animal a menos numa das listas). Por isso, as tabelas (seção H) exigem que todas as colunas tenham o mesmo tamanho.

````r
c(1, 2, 3, 4) + c(10, 20)      # recicla em silêncio: 1+10, 2+20, 3+10, 4+20
c(1, 2, 3, 4) + c(10, 20, 30)  # tamanhos incompatíveis: resultado com aviso (warning)
````

Como o vetor só guarda **um tipo de dado**, misturar tipos faz o R converter tudo para o tipo mais geral, em geral texto. Isso se chama **coerção**. É o mesmo problema da seção D: uma única célula com texto transforma a coluna inteira em texto.

````r
misturado <- c(98, 105, "não medido", 110)
misturado          # todos os valores viraram texto (note as aspas)
class(misturado)
````

## H. Estruturas de dados
Além do vetor, quatro estruturas aparecem com frequência na análise de dados:

| Estrutura | O que é | Tipos de dado |
|---|---|---|
| **vetor** | uma sequência de valores | um só tipo |
| **fator** (`factor`) | um vetor de categorias, com níveis definidos | categorias |
| **tabela** (`data.frame` ou `tibble`) | colunas de mesmo tamanho, lado a lado | cada coluna tem o seu tipo |
| **lista** (`list`) | uma coleção de objetos quaisquer | qualquer tipo, inclusive outras listas |

Também existe a **matriz** (`matrix`), uma tabela em que todos os valores têm o mesmo tipo, mais usada em cálculos numéricos e menos comum na análise de dados do dia a dia.

### Fatores: categorias com níveis
Variáveis categóricas, como grupo experimental, sexo ou dieta, costumam ser guardadas como **fatores**. Um fator tem **níveis** (as categorias possíveis) numa ordem definida. Essa ordem decide, por exemplo, a ordem dos grupos nos gráficos e qual grupo é a **referência** nos modelos estatísticos.

Por padrão, os níveis ficam em ordem alfabética, o que nem sempre faz sentido:

````r
grupo <- c("tratado", "controle", "tratado", "controle")

grupo_fator <- factor(grupo)
grupo_fator
levels(grupo_fator)   # ordem alfabética: "controle" antes de "tratado"

# Definindo a ordem dos níveis explicitamente
dieta <- factor(c("padrao", "hiperlipidica", "padrao"),
                levels = c("padrao", "hiperlipidica"))   # a dieta padrão é a referência
levels(dieta)
````

O `tibble` é a versão da tabela usada pelos pacotes do tidyverse: funciona como um `data.frame`, mas mostra o tipo de cada coluna e só as primeiras linhas, quando a tabela é grande.

## I. Dados arrumados (*tidy data*)
Uma mesma informação pode ser organizada em tabelas de formas muito diferentes, e a forma escolhida afeta tudo o que vem depois: o código, os gráficos e os testes. Uma forma padronizada de organizar tabelas para a análise foi proposta por **Hadley Wickham** no artigo *Tidy Data* (*Journal of Statistical Software*, 2014). Wickham é estatístico, cientista-chefe da Posit (a empresa do RStudio) e autor de vários pacotes que usamos nesta aula, como o `ggplot2`, o `dplyr` e o `tidyr`. O nome **tidyverse** vem justamente dessa ideia: é o "universo" de pacotes feitos para trabalhar com dados arrumados.

### As três regras
Um conjunto de dados está **arrumado** (*tidy*) quando:
1. **cada variável é uma coluna;**
2. **cada observação é uma linha;**
3. **cada valor ocupa uma célula.**

Aqui, **variável** é algo que se mede ou se registra (peso, sexo, dose, pressão), e **observação** é o conjunto de medidas feitas numa mesma unidade, numa mesma ocasião (por exemplo, um animal numa determinada semana).

> Essa é a formulação do livro *R for Data Science*, do mesmo autor. No artigo original, a terceira regra é "cada tipo de unidade observacional forma uma tabela": por exemplo, as características fixas dos animais (sexo, linhagem) numa tabela e as medidas repetidas (peso a cada semana) em outra. As ideias não são novas: Wickham as apresenta como a tradução, para a linguagem da estatística, da terceira forma normal de Codd, usada em bancos de dados.

Wickham resume a motivação parafraseando a abertura de *Anna Kariênina*, de Tolstói: os conjuntos de dados arrumados se parecem todos, mas cada conjunto desarrumado é desarrumado à sua maneira. Os problemas mais comuns que ele descreve são:
- **nomes de colunas que são valores**, e não nomes de variáveis (por exemplo, colunas `semana_1`, `semana_2`, `semana_3`);
- **várias variáveis numa só coluna** (por exemplo, uma coluna `grupo` com valores como `F_padrao`, que junta sexo e dieta);
- **variáveis guardadas tanto nas linhas quanto nas colunas**;
- **vários tipos de unidade observacional na mesma tabela**;
- **uma mesma unidade observacional espalhada por várias tabelas**.

### Um exemplo
O peso (g) de três ratos, medido em três semanas. Primeiro, no formato **largo**, como costuma ser digitado numa planilha, com uma coluna por semana:

````r
peso_largo <- data.frame(
  animal   = c("R1", "R2", "R3"),
  semana_1 = c(250, 262, 248),
  semana_2 = c(268, 280, 259),
  semana_3 = c(287, 301, 275)
)
peso_largo
````

A tabela é fácil de ler, mas não está arrumada: a **semana** é uma variável, e aqui seus valores (1, 2, 3) estão nos **nomes das colunas**. Além disso, o **peso** está espalhado por três colunas.

A mesma informação, no formato **longo** e **arrumado**, tem uma coluna para cada variável (`animal`, `semana`, `peso`) e uma linha para cada observação (um animal numa semana):

````r
peso_tidy <- data.frame(
  animal = rep(c("R1", "R2", "R3"), times = 3),   # os 3 animais, repetidos para cada semana
  semana = rep(1:3, each = 3),                     # cada semana, repetida para os 3 animais
  peso   = c(250, 262, 248,    # semana 1
             268, 280, 259,    # semana 2
             287, 301, 275)    # semana 3
)
peso_tidy
````

### Por que arrumar?
- **Cada coluna é um vetor** (seção F). As operações vetorizadas (seção G) agem sobre uma variável inteira de uma vez: `mean(peso_tidy$peso)` usa todos os pesos, enquanto, na tabela larga, os pesos estão divididos em três colunas.
- **As ferramentas do tidyverse esperam dados arrumados.** No `ggplot2`, cada eixo ou cor corresponde a uma coluna; no `dplyr`, filtrar e agrupar são operações sobre colunas. Com a tabela arrumada, perguntas como "qual é o peso médio em cada semana?" ou "qual foi o ganho de peso de cada animal?" viram poucas linhas de código.
- **Os modelos estatísticos também.** Uma fórmula como `peso ~ semana` só pode ser escrita se existirem as colunas `peso` e `semana`.

## 1. Os dados e o plano de análise
**Dados (simulados, criados para o workshop):** pressão arterial sistólica (mmHg) de 60 ratos hipertensos, tratados com um de dois fármacos (`A` ou `B`) em uma de três doses (1, 3 ou 10 mg/kg), com 10 animais por grupo.

**Plano de análise:**
- **Pergunta principal (confirmatória):** na dose de 1 mg/kg, a pressão arterial sistólica média difere entre os fármacos A e B?
- **Teste:** t de Welch, bilateral, com nível de significância de 5%, relatando a diferença entre as médias com intervalo de confiança de 95%.
- **Qualquer análise fora deste plano** será relatada como **exploratória**.

> Num estudo real, este plano seria definido e registrado publicamente antes da coleta dos dados (por exemplo, no OSF). Aqui, com dados simulados, ele serve de modelo.

## 3. Preparar os dados para a análise
Arrumar a tabela à mão, copiando e colando, funciona, mas não deixa registro e abre espaço para erros. Com código, são três passos, encadeados pelo **pipe** (`|>` ou `%>%`), que entrega o resultado de um passo como entrada do passo seguinte (leia como "e então"):
1. `pivot_longer()`: empilha as seis colunas numa só (`pas`: pressão arterial sistólica), guardando o nome de cada coluna de origem;
2. `separate_wider_delim()`: separa esse nome em fármaco e dose;
3. `mutate()`: transforma a dose em número.

````r
pressao_tidy <- pressao |>
  pivot_longer(everything(), names_to = "grupo", values_to = "pas") |>
  separate_wider_delim(grupo, delim = "_", names = c("farmaco", "dose")) |>
  mutate(dose = as.numeric(dose))
````

````r
pressao_tidy
````

Agora há uma linha por animal (60) e uma coluna por variável (`farmaco`, `dose`, `pas`). A tabela original não foi alterada, e cada passo da transformação está registrado no código.

## 4. Resumir por grupo com o `dplyr`
Com os dados arrumados, cada pergunta vira uma sequência de verbos: agrupar por fármaco e dose (`group_by()`), e então resumir cada grupo (`summarise()`) com o número de animais, a média e o desvio-padrão.

````r
pressao_tidy |>
  group_by(farmaco, dose) |>          # um grupo para cada combinação de fármaco e dose
  summarise(                          # calcula, para cada grupo:
    n = n(),                          #   o número de animais
    media = mean(pas),                #   a média da pressão
    desvio_pd = sd(pas),              #   o desvio-padrão da pressão
    .groups = "drop"                  # devolve uma tabela comum, sem os grupos
  )
````

## 5. Olhar os dados antes de testar
No `ggplot2`, um gráfico é montado em partes, somadas com `+`:
1. **dados e mapeamentos estéticos** (`aes()`): qual variável vai para o eixo x, qual vai para o y e qual define a cor;
2. **camadas**, cada uma com uma forma geométrica (*geom*): aqui, caixas (`geom_boxplot()`) e um ponto para cada animal (`geom_point()`);
3. **ajustes** de rótulos (`labs()`) e aparência (`theme_minimal()`).

Mostrar os pontos individuais, e não só as caixas, deixa visível cada animal: com 10 por grupo, dá para ver toda a variação.

````r
ggplot(pressao_tidy, aes(x = factor(dose), y = pas, fill = farmaco)) +   # dados e mapeamentos
  geom_boxplot() +                                                  # camada 1: boxplot
  geom_point(position = position_jitterdodge(jitter.width = 0.15),  # camada 2: um ponto por animal
             alpha = 0.6) +
  labs(x = "Dose (mg/kg)", y = "Pressão arterial sistólica (mmHg)",
       fill = "Fármaco") +                                          # rótulos
  theme_minimal()                                                   # aparência (minimal, classic, dark, light...)
````

## 6. Testar uma pergunta definida antes
**Na dose de 1 mg/kg, a pressão arterial média difere entre os fármacos?** Teste t de Welch para dois grupos independentes.

````r
pressao_tidy |>
  filter(dose == 1) |> # Filtre apenas tratamentos com dose 1 mg/kg
  t.test(pas ~ farmaco, data = _) # Compare a pressão entre os fármacos
````

Como definido no plano, relatamos a **diferença entre as médias e o seu intervalo de confiança**, não só o valor de p:

*Na dose de 1 mg/kg, a pressão média com o fármaco A foi cerca de 11 mmHg menor que com o fármaco B (IC 95%: 4 a 18 mmHg; p = 0,003).*

> **Pergunta para a turma:** o teste na dose de 10 mg/kg não estava no plano. Esse resultado poderia ser relatado da mesma forma que o da pergunta principal?

<details>
<summary><b>Resposta</b> (clique para abrir)</summary>

**Não.** Essa comparação não estava no plano: é uma **análise exploratória**. Ela pode sugerir hipóteses para um próximo estudo, mas não as confirma. Cada teste a mais aumenta a chance de um resultado "significativo" por acaso.

Se a pergunta também interessa nas outras doses, o caminho é **incluir essas comparações no plano desde o início** e **corrigir para comparações múltiplas** (por exemplo, pelo método de Holm).

Entretanto, a exploração também tem seu valor. No gráfico e nos resumos por grupo, a diferença entre A e B parece diminuir conforme a dose aumenta (cerca de 11, 7 e 5,5 mmHg). Isso não é uma conclusão, e sim uma **hipótese**, para ser testada num novo estudo, com um plano próprio: os mesmos dados não podem, ao mesmo tempo, sugerir uma hipótese e confirmá-la.
